```python

%cd /private/home/fcharton/ChromoBoot
import numpy as np
import time
from generate_data import convert
import os

path = '/private/home/fcharton/Amplitudes/ML/'

def get_loop_data(loop):
    if loop < 6:
        return convert(loop, path, "EZ_symb_new_norm", False)
    if loop == 6:
        return convert(loop, path, "EZ6_symb_new_norm", False)
    print("loop up to 6")
    return None

def query_loop(key,data):
    if key in data.keys():
        return data[key]
    return 0
    
def make_lists(loop):
    e3 = get_loop_data(loop-1)
    e4 = get_loop_data(loop)
    res = []
    for key,value in e4.items():
        nl = len(key)
        lst = []
        for j in range(nl-1):
            for k in range(j+1,nl):
                key2=key[:j]+key[j+1:k]+key[k+1:]
                lst.append(query_loop(key2,e3))
        res.append((lst,value))
    return res

print(make_lists(4))
```
