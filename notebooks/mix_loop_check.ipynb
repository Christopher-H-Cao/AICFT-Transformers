```python
import pandas as pd
import numpy as np
import json
import itertools
from itertools import islice
from matplotlib import pyplot as plt
mydir = '/home/garrett/work/mixed_loop'
```

```python
bad_6trim_array=[]
bad_6trim_df = pd.DataFrame(columns = ['word', 'tgt', 'pred'])
with open('/home/garrett/work/mixed_loop/checkpoint_6loop_compare_eval6/htcondor/dumped/debug/1333280/eval.valid.boots.199') as f:
    while True:
        next_n_lines = list(islice(f, 5))
        if not next_n_lines:
            break
        if next_n_lines[3][0]=='1': continue
        print(next_n_lines)
        word=next_n_lines[1][4:16]
        tgt=next_n_lines[2][4:-1]
        pred=next_n_lines[3][2:-1]
        data=[word, tgt, pred]
        bad_6trim_array.append(data)
bad_6trim_df = pd.DataFrame(bad_6trim_array, columns = ['word', 'tgt', 'pred'])
```

```python
bad_mix_array=[]
bad_mix_df = pd.DataFrame(columns = ['word', 'tgt', 'pred'])
with open('/home/garrett/work/mixed_loop/checkpoint_5loop6loopmix_eval6/htcondor/dumped/debug/1333300/eval.valid.boots.199') as f:
    while True:
        next_n_lines = list(islice(f, 5))
        if not next_n_lines:
            break
        if next_n_lines[3][0]=='1': continue
        word=next_n_lines[1][4:16]
        tgt=next_n_lines[2][4:-1]
        pred=next_n_lines[3][2:-1]
        data=[word, tgt, pred]
        bad_mix_array.append(data)
bad_mix_df = pd.DataFrame(bad_mix_array, columns = ['word', 'tgt', 'pred'])
```

```python
joint_df = pd.merge(bad_mix_df, bad_6trim_df, how ='inner', on =['word']).drop(columns=['tgt_y'])
joint_df = joint_df.rename(columns={'tgt_x': 'tgt', 'pred_x': 'pred_mix','pred_y': 'pred_6'})
pd.set_option('display.max_rows', 500)
print(joint_df) 
```

```python
left_df = pd.merge(bad_mix_df, bad_6trim_df, how ='left', on =['word'], indicator=True)
left_df = left_df[left_df["_merge"]=="left_only"].drop(columns=['tgt_y','pred_y','_merge'])
left_df = left_df.rename(columns={'tgt_x': 'tgt', 'pred_x': 'pred_mix'})
pd.set_option('display.max_rows', 500)
print(left_df) 
```

```python
right_df = pd.merge(bad_mix_df, bad_6trim_df, how ='right', on =['word'], indicator=True)
right_df = right_df[right_df["_merge"]=="right_only"].drop(columns=['tgt_x','pred_x','_merge'])
right_df = right_df.rename(columns={'tgt_y': 'tgt', 'pred_y': 'pred_6'})
pd.set_option('display.max_rows', 500)
print(right_df) 
```
